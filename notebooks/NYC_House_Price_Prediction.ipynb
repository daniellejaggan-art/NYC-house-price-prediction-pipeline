```python
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

categorical_cols = [
    "borough",
    "building_class_at_time_of_sale"
]

numerical_cols = [
    "gross_square_feet",
    "total_units",
    "year_built",
    "land_square_feet"
]

numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy= "median"))
])
```

```python
test_df = X_test.copy()
test_df["sale_price"] = Y_test
```

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy = "most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown ="ignore"))
])
```

```python
preprocessor = ColumnTransformer([
    ("numerical", numerical_pipeline, numerical_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])
```

```python
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])
```

```python
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
mae = mean_absolute_error(Y_test, Y_pred)

rsme = np.sqrt(mean_squared_error(Y_test, Y_pred))

r2 = r2_score(Y_test, Y_pred)

print("Mean Absolute Error (MAE):", mae)
print("Root Mean Squared Error (RSME):", rsme)
print("R2 Score:", r2)
```

```python
def load_property_data(engine):
    query = text("""
        SELECT
            borough,
            building_class_at_time_of_sale,
            gross_square_feet,
            total_units,
            year_built,
            land_square_feet
        FROM public.model_ready_sales;
    """)

    with engine.connect() as connection:
        property_data = pd.read_sql(query, connection)

    return property_data


property_data = load_property_data(engine)

print("Property data loaded successfully!")
print("Number of properties:", len(property_data))
print("Input columns:", property_data.columns.tolist())

display(property_data.head())
```

```python
def validate_property_data(property_data):
    required_columns = [
        "borough",
        "building_class_at_time_of_sale",
        "gross_square_feet",
        "total_units",
        "year_built",
        "land_square_feet"
    ]

   
    missing_columns = [
        col for col in required_columns
        if col not in property_data.columns
    ]

    if missing_columns:
        raise ValueError(
            f"Missing required columns: {missing_columns}"
        )

    if property_data.empty:
        raise ValueError("No property records were loaded.")

  
    duplicate_count = property_data.duplicated().sum()

    
    missing_values = property_data.isnull().sum()

    print("Validation checks completed.")
    print("Number of rows:", len(property_data))
    print("Duplicate rows:", duplicate_count)
    print("\nMissing values by column:")
    print(missing_values)

    return property_data


validated_data = validate_property_data(property_data)
```

```python
def generate_predictions(validated_data, model):
    predictions = model.predict(validated_data)

    results = validated_data.copy()
    results["predicted_sale_price"] = predictions

    return results


prediction_results = generate_predictions(
    validated_data,
    loaded_model
)

print("Predictions generated successfully!")
print("Number of predictions:", len(prediction_results))

display(prediction_results.head())
```

```python
def run_property_pipeline(engine, model):
    
    property_data = load_property_data(engine)

    
    validated_data = validate_property_data(property_data)

  
    results = generate_predictions(validated_data, model)

    print("\nPipeline completed successfully!")
    print("Total predictions:", len(results))

    return results


pipeline_results = run_property_pipeline(
    engine,
    loaded_model
)

display(pipeline_results.head())
```

```python

import os
from datetime import datetime

import joblib
import pandas as pd
import streamlit as st
import plotly.express as px
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

# --------------------------------------------------
# CONFIGURATION
# --------------------------------------------------

MODEL_PATH = "/Users/daniellejaggan/nyc_house_price_model.joblib"

FEATURES = [
    "borough",
    "building_class_at_time_of_sale",
    "gross_square_feet",
    "total_units",
    "year_built",
    "land_square_feet",
]

BOROUGHS = {
    "Manhattan": 1,
    "Bronx": 2,
    "Brooklyn": 3,
    "Queens": 4,
    "Staten Island": 5,
}

# Set these only after verifying they belong to the
# exact model saved at MODEL_PATH.
MODEL_METRICS = {
    "MAE": None,
    "RMSE": None,
    "R²": None,
}

# --------------------------------------------------
# PAGE AND LUXURY NAVY / GOLD THEME
# --------------------------------------------------

st.set_page_config(
    page_title="EstateIQ | NYC Property Intelligence",
    page_icon="🏙️",
    layout="wide",
    initial_sidebar_state="expanded",
)

st.markdown(
    """
    <style>
    .stApp {
        background-color: #071525;
        color: #F5F3ED;
    }

    [data-testid="stSidebar"] {
        background-color: #0B1D31;
        border-right: 1px solid #8F7445;
    }

    h1, h2, h3 {
        color: #E0C184 !important;
    }

    p, label, [data-testid="stMetricLabel"] {
        color: #D8DFE8 !important;
    }

    [data-testid="stMetricValue"] {
        color: #E0C184 !important;
    }

    div[data-testid="stForm"],
    div[data-testid="stVerticalBlockBorderWrapper"] {
        border: 1px solid #8F7445;
        border-radius: 12px;
        padding: 18px;
        background-color: #0D2035;
    }

    .hero {
        padding: 26px 30px;
        border-radius: 14px;
        border: 1px solid #8F7445;
        background: linear-gradient(115deg, #0D2035, #192A3B);
        margin-bottom: 22px;
    }

    .hero-label {
        color: #D5B779;
        text-transform: uppercase;
        letter-spacing: 3px;
        font-size: 12px;
    }

    .hero-subtitle {
        color: #D8DFE8;
        font-size: 16px;
    }

    .price-card {
        padding: 24px;
        border: 1px solid #B69A60;
        border-radius: 12px;
        background-color: #10243A;
        text-align: center;
    }

    .price-label {
        color: #D8DFE8;
        font-size: 14px;
    }

    .price-value {
        color: #E0C184;
        font-size: clamp(28px, 4vw, 42px);
        font-weight: 700;
        margin: 10px 0;
    }

    .stButton > button,
    .stFormSubmitButton > button {
        background-color: #C2A267;
        color: #071525;
        border: 0;
        border-radius: 8px;
        font-weight: 700;
        min-height: 42px;
        width: 100%;
    }

    .stButton > button:hover,
    .stFormSubmitButton > button:hover {
        background-color: #E0C184;
        color: #071525;
    }

    hr {
        border-color: #314258;
    }
    </style>
    """,
    unsafe_allow_html=True,
)

# --------------------------------------------------
# LOAD THE EXISTING MODEL
# --------------------------------------------------

@st.cache_resource
def load_model():
    return joblib.load(MODEL_PATH)


try:
    model = load_model()
except Exception as exc:
    st.error(f"Could not load the saved model: {exc}")
    st.stop()

# --------------------------------------------------
# DATABASE CONNECTION
# --------------------------------------------------

@st.cache_resource
def get_engine(database_url):
    return create_engine(
        database_url,
        pool_pre_ping=True,
    )


database_url = os.getenv("DATABASE_URL")
engine = None
db_error = None

if database_url:
    try:
        engine = get_engine(database_url)

        with engine.connect() as connection:
            connection.execute(text("SELECT 1"))

    except SQLAlchemyError as exc:
        db_error = str(exc)
        engine = None

# --------------------------------------------------
# CREATE PREDICTION HISTORY TABLE
# --------------------------------------------------

def initialize_history_table():
    if engine is None:
        return

    with engine.begin() as connection:
        connection.execute(text("""
            CREATE TABLE IF NOT EXISTS
            public.property_price_predictions (
                prediction_id BIGINT
                    GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
                predicted_at TIMESTAMPTZ
                    NOT NULL DEFAULT CURRENT_TIMESTAMP,
                borough INTEGER NOT NULL,
                building_class_at_time_of_sale TEXT NOT NULL,
                gross_square_feet DOUBLE PRECISION NOT NULL,
                total_units INTEGER NOT NULL,
                year_built INTEGER NOT NULL,
                land_square_feet DOUBLE PRECISION NOT NULL,
                predicted_sale_price DOUBLE PRECISION NOT NULL
            )
        """))


if engine is not None:
    try:
        initialize_history_table()
    except SQLAlchemyError as exc:
        st.warning(f"Prediction history setup failed: {exc}")

# --------------------------------------------------
# GET VALID BUILDING CLASSES FROM THE DATABASE
# --------------------------------------------------

@st.cache_data(ttl=300)
def get_building_classes():
    if engine is None:
        return []

    with engine.connect() as connection:
        result = connection.execute(text("""
            SELECT DISTINCT building_class_at_time_of_sale
            FROM public.model_ready_sales
            WHERE building_class_at_time_of_sale IS NOT NULL
            ORDER BY building_class_at_time_of_sale
        """))

        return [row[0] for row in result.fetchall()]


try:
    building_classes = get_building_classes()
except SQLAlchemyError:
    building_classes = []

# --------------------------------------------------
# SAVE AND RETRIEVE PREDICTIONS
# --------------------------------------------------

def save_prediction(features, predicted_price):
    if engine is None:
        raise RuntimeError(
            "PostgreSQL is not connected. "
            "Configure DATABASE_URL first."
        )

    with engine.begin() as connection:
        connection.execute(
            text("""
                INSERT INTO public.property_price_predictions (
                    borough,
                    building_class_at_time_of_sale,
                    gross_square_feet,
                    total_units,
                    year_built,
                    land_square_feet,
                    predicted_sale_price
                )
                VALUES (
                    :borough,
                    :building_class,
                    :gross_sqft,
                    :units,
                    :year_built,
                    :land_sqft,
                    :predicted_price
                )
            """),
            {
                "borough": features["borough"],
                "building_class": features[
                    "building_class_at_time_of_sale"
                ],
                "gross_sqft": features["gross_square_feet"],
                "units": features["total_units"],
                "year_built": features["year_built"],
                "land_sqft": features["land_square_feet"],
                "predicted_price": float(predicted_price),
            },
        )


def load_prediction_history(limit=20):
    if engine is None:
        return pd.DataFrame()

    with engine.connect() as connection:
        return pd.read_sql(
            text("""
                SELECT
                    prediction_id,
                    predicted_at,
                    borough,
                    building_class_at_time_of_sale,
                    gross_square_feet,
                    total_units,
                    year_built,
                    land_square_feet,
                    predicted_sale_price
                FROM public.property_price_predictions
                ORDER BY predicted_at DESC
                LIMIT :limit
            """),
            connection,
            params={"limit": limit},
        )


# --------------------------------------------------
# FEATURE IMPORTANCE
# --------------------------------------------------

def get_feature_importance(loaded_model):
    estimator = loaded_model

    # Handles a scikit-learn Pipeline whose final step
    # is the fitted estimator.
    if hasattr(loaded_model, "steps"):
        estimator = loaded_model.steps[-1][1]

    if not hasattr(estimator, "feature_importances_"):
        return None

    importances = estimator.feature_importances_

    # Prefer transformed names if preprocessing provides them.
    if hasattr(loaded_model, "steps"):
        try:
            names = loaded_model[:-1].get_feature_names_out()
        except Exception:
            names = None
    else:
        names = None

    if names is None or len(names) != len(importances):
        names = getattr(
            estimator,
            "feature_names_in_",
            None,
        )

    if names is None or len(names) != len(importances):
        names = [
            f"Feature {i + 1}"
            for i in range(len(importances))
        ]

    importance_df = pd.DataFrame({
        "Feature": names,
        "Importance": importances,
    })

    return importance_df.sort_values(
        "Importance",
        ascending=True,
    )


# --------------------------------------------------
# SIDEBAR
# --------------------------------------------------

with st.sidebar:
    st.markdown("## ✦ EstateIQ")
    st.caption("NYC PROPERTY INTELLIGENCE")
    st.divider()

    page = st.radio(
        "NAVIGATION",
        [
            "Property Valuation",
            "Model Insights",
            "Prediction History",
        ],
    )

    st.divider()
    st.caption("REAL DATA. INFORMED DECISIONS.")

    if engine is not None:
        st.success("PostgreSQL connected")
    else:
        st.warning("PostgreSQL not connected")

# --------------------------------------------------
# HEADER
# --------------------------------------------------

st.markdown(
    """
    <div class="hero">
        <div class="hero-label">EstateIQ · Property Intelligence</div>
        <h1 style="margin-bottom: 8px;">
            NYC Property Price Predictor
        </h1>
        <div class="hero-subtitle">
            Explore estimated property values using your
            trained Random Forest machine learning model.
        </div>
    </div>
    """,
    unsafe_allow_html=True,
)

# --------------------------------------------------
# PAGE 1: PROPERTY VALUATION
# --------------------------------------------------

if page == "Property Valuation":

    st.subheader("Property Details")
    st.write(
        "Enter the property's characteristics to generate "
        "an estimated sale price."
    )

    if not building_classes:
        st.warning(
            "Building classes could not be loaded from "
            "PostgreSQL. Connect to your database before "
            "making predictions."
        )

    with st.form("valuation_form"):

        col1, col2 = st.columns(2)

        with col1:
            borough_name = st.selectbox(
                "Borough",
                list(BOROUGHS.keys()),
            )

            gross_sqft = st.number_input(
                "Gross Square Feet",
                min_value=1,
                max_value=10_000_000,
                value=1500,
                step=100,
            )

            year_built = st.number_input(
                "Year Built",
                min_value=1800,
                max_value=datetime.now().year,
                value=2000,
            )

        with col2:
            building_class = st.selectbox(
                "Building Class",
                building_classes,
                disabled=not building_classes,
            )

            total_units = st.number_input(
                "Total Units",
                min_value=1,
                max_value=100_000,
                value=1,
            )

            land_sqft = st.number_input(
                "Land Square Feet",
                min_value=1,
                max_value=100_000_000,
                value=2000,
                step=100,
            )

        submitted = st.form_submit_button(
            "✦  Predict Sale Price",
            disabled=not building_classes,
        )

    if submitted:
        property_data = pd.DataFrame([{
            "borough": BOROUGHS[borough_name],
            "building_class_at_time_of_sale": building_class,
            "gross_square_feet": float(gross_sqft),
            "total_units": int(total_units),
            "year_built": int(year_built),
            "land_square_feet": float(land_sqft),
        }], columns=FEATURES)

        try:
            # Do not manually encode categorical values here.
            # The saved pipeline must handle its own preprocessing.
            predicted_price = float(
                model.predict(property_data)[0]
            )

            if not pd.notna(predicted_price):
                raise ValueError("Model returned an invalid prediction.")

            if predicted_price < 0:
                st.warning(
                    "The model returned a negative estimate. "
                    "Review the model and its training data."
                )

            st.markdown(
                f"""
                <div class="price-card">
                    <div class="price-label">
                        ESTIMATED SALE PRICE
                    </div>
                    <div class="price-value">
                        ${predicted_price:,.0f}
                    </div>
                    <div class="price-label">
                        Random Forest · Model estimate
                    </div>
                </div>
                """,
                unsafe_allow_html=True,
            )

            st.write("")
            st.caption(
                "This estimate is not a formal appraisal or "
                "a guarantee of market value."
            )

            if engine is not None:
                try:
                    save_prediction(
                        property_data.iloc[0].to_dict(),
                        predicted_price,
                    )
                    st.success(
                        "Prediction saved to PostgreSQL history."
                    )
                except Exception as exc:
                    st.warning(
                        f"Prediction succeeded, but saving history "
                        f"failed: {exc}"
                    )
            else:
                st.info(
                    "Prediction was generated, but history was "
                    "not saved because PostgreSQL is unavailable."
                )

        except Exception as exc:
            st.error(
                "Prediction failed. The input types or feature "
                f"encoding may not match the saved model: {exc}"
            )

# --------------------------------------------------
# PAGE 2: MODEL INSIGHTS
# --------------------------------------------------

elif page == "Model Insights":

    st.subheader("Model Performance")

    st.write(
        "These metrics should come from the held-out test set "
        "used to evaluate the exact saved model."
    )

    metric_cols = st.columns(3)

    for column, name in zip(
        metric_cols,
        ["MAE", "RMSE", "R²"],
    ):
        value = MODEL_METRICS[name]

        with column:
            if value is None:
                st.metric(name, "Not verified")
            elif name == "R²":
                st.metric(name, f"{value:.4f}")
            else:
                st.metric(name, f"${value:,.0f}")

    if all(
        MODEL_METRICS[name] is None
        for name in ["MAE", "RMSE", "R²"]
    ):
        st.info(
            "Add the verified evaluation results from your model "
            "training notebook to display the actual metrics."
        )

    st.divider()
    st.subheader("Feature Importance")

    importance_df = get_feature_importance(model)

    if importance_df is not None:
        importance_df["Display Feature"] = (
            importance_df["Feature"]
            .astype(str)
            .str.replace(
                "num__",
                "",
                regex=False,
            )
            .str.replace(
                "cat__",
                "",
                regex=False,
            )
        )

        chart = px.bar(
            importance_df,
            x="Importance",
            y="Display Feature",
            orientation="h",
            title="Relative Feature Importance",
            color_discrete_sequence=["#C2A267"],
        )

        chart.update_layout(
            template="plotly_dark",
            paper_bgcolor="#0D2035",
            plot_bgcolor="#0D2035",
            font_color="#F5F3ED",
            xaxis_title="Importance",
            yaxis_title="",
            margin=dict(l=20, r=20, t=60, b=20),
        )

        st.plotly_chart(chart, use_container_width=True)

        st.caption(
            "Feature importance describes how the fitted model "
            "uses features; it does not establish causation."
        )
    else:
        st.info(
            "The saved estimator does not expose feature_importances_. "
            "We can inspect the model and choose an appropriate "
            "explanation method if necessary."
        )

# --------------------------------------------------
# PAGE 3: PREDICTION HISTORY
# --------------------------------------------------

elif page == "Prediction History":

    st.subheader("Recent Predictions")

    if engine is None:
        st.warning(
            "Connect PostgreSQL to view saved prediction history."
        )
    else:
        try:
            history = load_prediction_history()

            if history.empty:
                st.info(
                    "No predictions have been saved yet. "
                    "Generate a prediction to begin your history."
                )
            else:
                history["predicted_sale_price"] = (
                    history["predicted_sale_price"]
                    .map(lambda price: f"${price:,.0f}")
                )

                st.dataframe(
                    history,
                    use_container_width=True,
                    hide_index=True,
                )

                st.download_button(
                    "Download Prediction History CSV",
                    data=history.to_csv(index=False).encode("utf-8"),
                    file_name="estateiq_prediction_history.csv",
                    mime="text/csv",
                )

        except Exception as exc:
            st.error(f"Could not load prediction history: {exc}")

# --------------------------------------------------
# FOOTER
# --------------------------------------------------

st.divider()

st.caption(
    "EstateIQ | NYC Property Intelligence | "
    "CS4998 Capstone Project"
)
```
